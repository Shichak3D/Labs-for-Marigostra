```python
from gdown import download
from gensim.downloader import load
from gensim.models import KeyedVectors
```

```python
model_path = r"model.bin"
```

```python
# model = KeyedVectors.load(model_path)
model = KeyedVectors.load_word2vec_format(
    model_path,
    binary=True
)
```

```python
word = "книга_NOUN"

positive = ["свадьба_NOUN"]
negative = ["любовь_NOUN"]
```

```python
def find_10_similarity_words(positive: list[str], negative: list[str]):
    global model

    result = model.most_similar(
        positive=positive,
        negative=negative,
        topn=10
    )

    return result
```

```python
similarity_words = find_10_similarity_words(positive, negative)
print(similarity_words)
```
