```python
from transformers import AutoModelForCausalLM, AutoTokenizer
import torch
```

```python
save_dir = "./rugpt"

device = "cuda" if torch.cuda.is_available() else "cpu"
print(device)

model_name = "ai-forever/rugpt3large_based_on_gpt2"
tokenizer = AutoTokenizer.from_pretrained(save_dir)
model = AutoModelForCausalLM.from_pretrained(save_dir).to(device)
```

```python
def generate(
    model, tok, text,
    do_sample=True, max_length=100, repetition_penalty=10.0,
    top_k=5, top_p=0.95, temperature=1,
    num_beams=5,
    no_repeat_ngram_size=3
):

    global device

    input_ids = tok.encode(text, return_tensors="pt").to(device)
    print(input_ids)
    print(model.generate.__globals__['__file__'])
    out = model.generate(
        input_ids,
        max_length=max_length,
        repetition_penalty=repetition_penalty,
        do_sample=do_sample,
        top_k=top_k,
        top_p=top_p,
        temperature=temperature,
        num_beams=num_beams,
        no_repeat_ngram_size=no_repeat_ngram_size
    )
    return list(map(tok.decode, out))
```

```python
prompt = "Если пройти по Томску"
print(generate(
    model, tokenizer, prompt
)[0])
```
