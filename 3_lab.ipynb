```python
import torch
from transformers import pipeline
```

```python
if torch.cuda.is_available():
    device = 0
else:
    device = -1
print(f"device:{device}")
```

```python
torch.__version__
```

```python
unmasker = pipeline(
    "fill-mask",
    model="ai-forever/ruBert-base",
    device=device,
    torch_dtype=torch.float16
)
```

```python
save_dir = "./bert_model"

unmasker.model.save_pretrained(save_dir)
unmasker.tokenizer.save_pretrained(save_dir)
```

```python
text = "Они [MASK] много еды в магазине и пошел в гости"

results = unmasker(text)

count_words = 10
for result in results[:count_words]:
    print(f"Слово: {result['token_str']}; вероятность слова:{result['score']}")
```
